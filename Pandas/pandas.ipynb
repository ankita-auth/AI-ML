# Series

```python
s=pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#index based access
print(s[0])
print(s[2])
```

```python
#label
s=pd.Series([23,24,25,26,"abc"], index=["Adam","Jibitesh","Ankita","Lucky",25])
print(s)

print(s["Adam"])
print(s["Lucky"])
#print(s.[0])  #Error
print(s.iloc[0])
print(s.index)

#NOTE:To get the dtype as fully  str or int , the whole list should be homogenous
#otherwise it will give an object
```

```python
#Vectorised Operation
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])

print(s1+s2)

#Mutable(Immutable Size)
s1[0]=100
changed_s1=s1.drop(0)

print(s1)
print(changed_s1)
```

```python
#Backticks
df1=pd.DataFrame([["Adam",23],["Bob",24],["Eve",25]], columns=["Name wame","Age"])
print(df)
df1.query("`Name wame`=='Adam'")
```

```python
#python vars
aqi_val=100
df.query("aqi>@aqi_val & temperature>30" )
```

```python
#Query-copy[by default it returns copy]
a=df.query("aqi>@aqi_val & temperature>30" ).copy()
```

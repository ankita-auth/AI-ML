# Melt & Pivot

```python
df=pd.DataFrame({
    "country":["USA","USA","India","India"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22],
})
df
```

```python
melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="metrics",
    value_name="value"
)
melted_df
```

```python
original_df=melted_df.pivot(
    index=["country","year"],
    columns="metrics",
    values="value"

)
original_df
```

# Merging & Joining Data

```python
df_customers=pd.DataFrame({
    "customer_id":[1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
})
df_orders=pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})
    
```

```python
df_customers
```

```python
df_orders
```

```python
#INNER JOIN

pd.merge(df_customers,df_orders,on="customer_id") #Only matching value will be shown
```

```python
#LEFT JOIN
pd.merge(df_customers,df_orders,on="customer_id", how="left") #Keeping every value of left table
```

```python
#RIGHT JOIN
pd.merge(df_customers,df_orders,on="customer_id", how="right") #Keeping every value of right table
```

```python
#OUTER JOIN

pd.merge(df_customers,df_orders,on="customer_id",how="outer") #Valuee of both table will be shown
```

# Data Concatenation

```python
df1=pd.DataFrame({
    "id":[1,2,3],
    "name":["Adam","Eve","Bob"]
})
df2=pd.DataFrame({
    "id":[4,5,6],
    "name":["Charlie","Fester","Gilian"]
})
#row-wise
pd.concat([df1,df2]) 
```

```python
pd.concat([df2,df1])
```

```python
pd.concat([df2,df1],ignore_index=True)
```

```python
#column-wise
pd.concat([df1,df2],axis=1) 
```

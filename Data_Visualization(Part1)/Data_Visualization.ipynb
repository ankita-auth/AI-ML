```python
#Real Data


#Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

#Plot
plt.plot(years,oscar_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
#Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception","Hugo", "Lincoln"]
non_oscar_revenue=[378, 2788, 879, 185, 275] # in $M

# plt.plot(years, oscar_revenue,"o--r", label="Oscar Movies")
plt.plot(years, oscar_revenue,color="red",marker="<",linestyle="--",linewidth=5, label="Oscar Movies")
plt.plot(years, non_oscar_revenue,"<-g", label="Non-Oscar Movies")
plt.title("Oscar Movies VS Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend(loc="upper right")
```

```python
#Tight Layout, SaveFig

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232] # in $M

plt.style.use("classic")
plt.plot(movies,revenue)
plt.title("Movies Revenue")
plt.xlabel("Movies")
plt.ylabel("Revenue ($M)")
plt.tight_layout()
plt.savefig("final_plot.png")
plt.show()
```

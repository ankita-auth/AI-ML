```python
# Task
import numpy as np
import matplotlib.pyplot as  plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
cities = ["New York", "London", "Delhi", "Tokyo"]
temperatures = [
    [22, 23, 21, 24, 25],  # New York
    [18, 19, 17, 20, 21],  # London
    [30, 32, 31, 33, 34],  # Delhi
    [25, 26, 24, 27, 28],  # Tokyo
]

fig,ax=plt.subplots(2,2,figsize=(5,5))

c=0
for i in range(2):
    for j in range(2):
        ax[i][j].plot(days,temperatures[c],marker='o')
        ax[i][j].set_title(cities[c])
        ax[i][j].grid()
        ax[i][j].set_xlabel("Days")
        ax[i][j].set_ylabel("Temperatures")
        c+=1


fig.suptitle("Temperatures in Cities  over the week")
fig.tight_layout()
```

```python
#Box Plot
import matplotlib.pyplot as plt

data=[7,8,5,6,9,4,10,12,15]

plt.boxplot(data)
plt.grid(True)
```

```python
#Outlier->100

data=[7,8,5,6,9,4,10,12,15,100]

plt.boxplot(data)
plt.grid(True)
```

```python
#Multiple Dataset
import numpy as np
group1=np.random.normal(50,10,100)
group2=np.random.normal(60,15,100)
#plt.boxplot([group1,group2], labels=["group1","group2"])->default=1,2
plt.boxplot([group1,group2], tick_labels=["group1","group2"], showmeans=True)
plt.grid(True)
plt.show()
```

## Horizontal Box Plots

```python
data=[7,8,5,6,9,4,10,12,15]

plt.boxplot(data, vert=False, showmeans=True,whis=2.0)
plt.grid(True)
```
